```python
#Exercício 21

def busca_binaria(lista, elemento):
    inicio = 0
    fim = len(lista) - 1

    while inicio <= fim:
        meio = (inicio + fim) // 2

        if lista[meio] == elemento:
            return meio
        elif lista[meio] < elemento:
            inicio = meio + 1
        else:
            fim = meio - 1

    return -1


lista = [1, 3, 5, 7, 9, 11, 13]
elemento = int(input("Digite o número que deseja buscar: "))

resultado = busca_binaria(lista, elemento)

if resultado != -1:
    print(f"Elemento encontrado na posição {resultado}.")
else:
    print("Elemento não encontrado.")
```

```python
#Exercício 22

def criptografar_cesar(mensagem, deslocamento):
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():

            # Define o início do alfabeto
            inicio = ord("A") if caractere.isupper() else ord("a")

            # Aplica o deslocamento
            novo_caractere = chr(
                (ord(caractere) - inicio + deslocamento) % 26 + inicio
            )

            resultado += novo_caractere
        else:
            resultado += caractere

    return resultado


mensagem = input("Digite a mensagem: ")
deslocamento = int(input("Digite o deslocamento: "))

print("Mensagem criptografada:")
print(criptografar_cesar(mensagem, deslocamento))
```

```python
#Exercício 23

def descriptografar_cesar(mensagem, deslocamento):

    # Para descriptografar, utilizamos o deslocamento negativo
    resultado = ""

    for caractere in mensagem:
        if caractere.isalpha():
            inicio = ord("A") if caractere.isupper() else ord("a")

            novo_caractere = chr(
                (ord(caractere) - inicio - deslocamento) % 26 + inicio
            )

            resultado += novo_caractere
        else:
            resultado += caractere

    return resultado


mensagem = input("Digite a mensagem criptografada: ")
deslocamento = int(input("Digite o deslocamento utilizado: "))

print("Mensagem descriptografada:")
print(descriptografar_cesar(mensagem, deslocamento))
```

```python
#Exercício 24

def validar_cpf(cpf):
    # Remove pontos e traços
    cpf = "".join(filter(str.isdigit, cpf))

    # Verifica se possui 11 dígitos ou se todos são iguais
    if len(cpf) != 11 or cpf == cpf[0] * 11:
        return False

    # Calcula o primeiro dígito verificador
    soma = sum(int(cpf[i]) * (10 - i) for i in range(9))
    digito1 = (soma * 10 % 11) % 10

    # Calcula o segundo dígito verificador
    soma = sum(int(cpf[i]) * (11 - i) for i in range(10))
    digito2 = (soma * 10 % 11) % 10

    # Compara os dígitos calculados
    return cpf[-2:] == f"{digito1}{digito2}"


cpf = input("Digite o CPF: ")

if validar_cpf(cpf):
    print("CPF válido.")
else:
    print("CPF inválido.")
```

```python
#Exercício 31

def bubble_sort(lista):
    tamanho = len(lista)

    # Percorre a lista diversas vezes
    for i in range(tamanho):

        for j in range(0, tamanho - i - 1):

            # Troca os elementos caso estejam fora de ordem
            if lista[j] > lista[j + 1]:
                lista[j], lista[j + 1] = lista[j + 1], lista[j]

    return lista


numeros = [64, 34, 25, 12, 22, 11, 90]

print("Lista ordenada:", bubble_sort(numeros))
```

```python
#Exercício 32

tabuleiro = [" "] * 9


def mostrar_tabuleiro():
    print(f"{tabuleiro[0]} | {tabuleiro[1]} | {tabuleiro[2]}")
    print("--+---+--")
    print(f"{tabuleiro[3]} | {tabuleiro[4]} | {tabuleiro[5]}")
    print("--+---+--")
    print(f"{tabuleiro[6]} | {tabuleiro[7]} | {tabuleiro[8]}")


def verificar_vencedor(jogador):
    combinacoes = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],
        [0, 3, 6], [1, 4, 7], [2, 5, 8],
        [0, 4, 8], [2, 4, 6]
    ]

    for combinacao in combinacoes:
        if all(tabuleiro[posicao] == jogador for posicao in combinacao):
            return True

    return False


jogador = "X"

for rodada in range(9):

    mostrar_tabuleiro()

    while True:
        posicao = int(
            input(f"Jogador {jogador}, escolha uma posição de 1 a 9: ")
        ) - 1

        if 0 <= posicao <= 8 and tabuleiro[posicao] == " ":
            break

        print("Posição inválida. Tente novamente.")

    tabuleiro[posicao] = jogador

    if verificar_vencedor(jogador):
        mostrar_tabuleiro()
        print(f"Jogador {jogador} venceu!")
        break

    jogador = "O" if jogador == "X" else "X"

else:
    mostrar_tabuleiro()
    print("Empate!")
```

```python
#Exercício 48

def insertion_sort(lista):

    # Percorre a lista a partir do segundo elemento
    for i in range(1, len(lista)):

        chave = lista[i]
        j = i - 1

        # Move os elementos maiores para a direita
        while j >= 0 and lista[j] > chave:
            lista[j + 1] = lista[j]
            j -= 1

        lista[j + 1] = chave

    return lista


numeros = [12, 11, 13, 5, 6]

print("Lista ordenada:", insertion_sort(numeros))
```
